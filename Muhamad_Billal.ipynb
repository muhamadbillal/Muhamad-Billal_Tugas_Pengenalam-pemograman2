```python
# konversi_ipk.py - Pemroses Transkrip Nilai Akademik

# Contoh input: Budi Santoso, 13524001, 80.5, 90.0, 78.25
data = input("Masukkan data (Nama,NIM,Tugas,UTS,UAS): ")

# 1. Pisahkan dengan split(",")
nama, nim, nilai_tugas, nilai_uts, nilai_uas = data.split(",")

# Bersihkan spasi dan konversi ke float
nama = nama.strip()
nim = nim.strip()
nilai_tugas = float(nilai_tugas)
nilai_uts = float(nilai_uts)
nilai_uas = float(nilai_uas)

# 2. Hitung Nilai Akhir dengan bobot
bobot_tugas = 0.20
bobot_uts = 0.35
bobot_uas = 0.45

nilai_akhir = (nilai_tugas * bobot_tugas) + (nilai_uts * bobot_uts) + (nilai_uas * bobot_uas)

# 3. Tampilkan dalam 1 frame kotak
print("-" * 42)
print(f"| {'TRANSKRIP NILAI AKADEMIK':^38} |")
print("-" * 42)
print(f"| {'Nama':<12}: {nama:<23} |")
print(f"| {'NIM':<12}: {nim:<23} |")
print("-" * 42)
print(f"| {'Nilai Tugas':<12}: {nilai_tugas:>6.1f}  ({bobot_tugas:.0%})      |")
print(f"| {'Nilai UTS':<12}: {nilai_uts:>6.1f}  ({bobot_uts:.0%})      |")
print(f"| {'Nilai UAS':<12}: {nilai_uas:>6.1f}  ({bobot_uas:.0%})      |")
print("-" * 42)
print(f"| {'Nilai Akhir':<12}: {nilai_akhir:>6.1f}                |")
print("-" * 42)
```
